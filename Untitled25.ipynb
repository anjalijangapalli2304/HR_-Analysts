```python
import pandas as pd

df = pd.read_csv("/content/Palo Alto Networks (1).csv")

print("Upload Successful!")
print(df.shape)
display(df.head())
```

```python
import numpy as np

# Feature Engineering

# Avoid division by zero
years = df['YearsAtCompany'].replace(0, np.nan)

# 1. Promotion Gap Ratio
df['PromotionGapRatio'] = (
    df['YearsSinceLastPromotion'] / years
)

# 2. Role Stagnation Index
df['RoleStagnationIndex'] = (
    df['YearsInCurrentRole'] / years
)

# 3. Training Intensity Score
df['TrainingIntensityScore'] = (
    df['TrainingTimesLastYear'] / years
)# 4. Manager Stability Indicator
df['ManagerStabilityIndicator'] = (
    df['YearsWithCurrManager'] / years
)

# Replace infinite values if any
df.replace([np.inf, -np.inf], np.nan, inplace=True)

print("Feature Engineering Completed Successfully!")

display(df[[
    'YearsAtCompany',
    'YearsSinceLastPromotion',
    'YearsInCurrentRole',
    'YearsInCurrentRole',
    'TrainingTimesLastYear',
    'YearsWithCurrManager',
    'PromotionGapRatio',
    'RoleStagnationIndex',
    'TrainingIntensityScore',
    'ManagerStabilityIndicator'
]].head())
```

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np

# Select numerical career-related columns
num_features = [
    'Age',
    'JobLevel',
    'TotalWorkingYears',
    'YearsAtCompany',
    'YearsInCurrentRole',
    'YearsSinceLastPromotion',
    'YearsWithCurrManager',
    'TrainingTimesLastYear',
    'PromotionGapRatio',
    'RoleStagnationIndex',
    'TrainingIntensityScore',
    'ManagerStabilityIndicator'
]

# Select categorical columns
cat_features = ['Department', 'JobRole']

# Prepare numerical data
X_num = df[num_features].copy()
X_num = X_num.replace([np.inf, -np.inf], np.nan)
X_num = X_num.fillna(X_num.median())

# Prepare categorical data
X_cat = df[cat_features].copy()
for col in cat_features:
    X_cat[col] = X_cat[col].fillna(X_cat[col].mode()[0])

# Encode categorical columns
X_cat_encoded = pd.get_dummies(
    X_cat, drop_first=False, dtype=int
)

# Combine features
X = pd.concat([X_num, X_cat_encoded], axis=1)

# Normalize numerical features
scaler = StandardScaler()
X_scaled = X.copy()
X_scaled[num_features] = scaler.fit_transform(X[num_features])

print("Data Preprocessing Completed!")
print("Original Dataset Shape:", df.shape)
print("Processed Dataset Shape:", X_scaled.shape)

display(X_scaled.head())
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Existing dataset
data = df.copy()

# Remove target column (if present)
data = data.drop(columns=['Attrition'], errors='ignore')

# Remove constant columns
data = data.loc[:, data.nunique() > 1]

# Separate numerical and categorical columns
num_cols = data.select_dtypes(include=['number']).columns
cat_cols = data.select_dtypes(exclude=['number']).columns

# Preprocessing
num_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

cat_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('num', num_pipe, num_cols),
    ('cat', cat_pipe, cat_cols)
])

X = preprocessor.fit_transform(data)

print("Preprocessing completed!")
print("Final dataset shape:", X.shape)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Select career progression features
features = [
    'YearsAtCompany',
    'YearsInCurrentRole',
    'YearsSinceLastPromotion',
    'YearsWithCurrManager',
    'TotalWorkingYears',
    'TrainingTimesLastYear',
    'JobLevel'
]

# Prepare data
X_kmeans = df[features].copy()

# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_kmeans)

# Find best number of clusters
for k in range(2, 7):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    print("Clusters:", k, "| Silhouette Score:", round(score, 3))
# Final K-Means model
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Career_Cluster'] = kmeans.fit_predict(X_scaled)

print("\nK-Means Clustering Completed!")
print(df['Career_Cluster'].value_counts().sort_index())
```

```python
# 1. Cluster-wise employee count
print("Cluster Distribution:")
print(df['Career_Cluster'].value_counts().sort_index())

# 2. Cluster Profile
cluster_profile = df.groupby('Career_Cluster')[
    ['YearsAtCompany',
     'YearsInCurrentRole',
     'YearsSinceLastPromotion',
     'YearsWithCurrManager',
     'TotalWorkingYears',
     'TrainingTimesLastYear',
     'JobLevel']
].mean().round(2)

print("\nCluster Profile:")
display(cluster_profile)

# 3. Attrition rate in each cluster
print("\nCluster-wise Attrition Rate:")
attrition_analysis = df.groupby('Career_Cluster')['Attrition'].mean() * 100
display(attrition_analysis.round(2))

# 4. Cluster visualization
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Career_Cluster')
plt.title('Career Progression Cluster Distribution')
plt.xlabel('Cluster')
plt.ylabel('Number of Employees')
plt.show()
```

```python
# Cluster-wise Attrition Analysis

attrition_result = df.groupby('Career_Cluster')['Attrition'].agg(
    Total_Employees='count',
    Attrition_Count='sum',
    Attrition_Rate='mean'
)

attrition_result['Attrition_Rate'] *= 100

display(attrition_result.round(2))

# Visualization
plt.figure(figsize=(8,5))

sns.barplot(
    x=attrition_result.index,
    y=attrition_result['Attrition_Rate']
)

plt.title('Attrition Rate Across Career Clusters')
plt.xlabel('Career Cluster')
plt.ylabel('Attrition Rate (%)')
plt.show()
```

```python
# Career Progression Analysis

career_analysis = df.groupby('Career_Cluster')[
    ['YearsAtCompany',
     'YearsInCurrentRole',
     'YearsSinceLastPromotion',
     'YearsWithCurrManager',
     'JobLevel']
].mean().round(2)

print("Career Progression Analysis:")
display(career_analysis)

# Visualization
plt.figure(figsize=(9,5))

sns.barplot(
    data=df,
    x='Career_Cluster',
    y='YearsSinceLastPromotion',
    errorbar=None
)

plt.title('Promotion Gap Across Career Clusters')
plt.xlabel('Career Cluster')
plt.ylabel('Average Years Since Last Promotion')
plt.show()
```

```python
# STEP 7: K-Means Clustering

import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Select career-related features
features = [
    'JobLevel',
    'TotalWorkingYears',
    'YearsAtCompany',
    'YearsInCurrentRole',
    'YearsSinceLastPromotion',
    'YearsWithCurrManager',
    'MonthlyIncome',
    'PercentSalaryHike',
    'PerformanceRating'
]

# Prepare data
X = df[features].copy()

# Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Find best number of clusters
scores = {}

for k in range(2, 7):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X_scaled)
    scores[k] = silhouette_score(X_scaled, labels)

print("Silhouette Scores:")
print(scores)

# Train final K-Means model
best_k = max(scores, key=scores.get)

kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

df['Career_Cluster'] = kmeans.fit_predict(X_scaled)

print("\nBest Number of Clusters:", best_k)
print("\nCluster Distribution:")
print(df['Career_Cluster'].value_counts().sort_index())
```

```python
cluster_summary = df.groupby('Career_Cluster').agg(
    Employee_Count=('Career_Cluster', 'count'),
    Average_Job_Level=('JobLevel', 'mean'),
    Average_Experience=('TotalWorkingYears', 'mean'),
    Average_Company_Years=('YearsAtCompany', 'mean'),
    Average_Role_Years=('YearsInCurrentRole', 'mean'),
    Average_Promotion_Gap=('YearsSinceLastPromotion', 'mean'),
    Average_Training=('TrainingTimesLastYear', 'mean')
).round(2)

print("CAREER CLUSTER SUMMARY")
display(cluster_summary)

# Visualize cluster profiles
cluster_summary[
    ['Average_Company_Years',
     'Average_Role_Years',
     'Average_Promotion_Gap']
].plot(kind='bar', figsize=(10,6))

plt.title('Career Progression Across Clusters')
plt.xlabel('Career Cluster')
plt.ylabel('Average Years')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# STEP 9: Promotion Gap Analysis

promotion_analysis = df.groupby('Career_Cluster').agg(
    Total_Employees=('Career_Cluster', 'size'),
    Average_Years_Since_Promotion=('YearsSinceLastPromotion', 'mean'),
    Average_Years_In_Current_Role=('YearsInCurrentRole', 'mean'),
    Average_Years_At_Company=('YearsAtCompany', 'mean')
).round(2)

print("PROMOTION GAP ANALYSIS")
display(promotion_analysis)

# Promotion Gap Graph
plt.figure(figsize=(9,5))

sns.barplot(
    data=df,
    x='Career_Cluster',
    y='YearsSinceLastPromotion',
    errorbar=None
)

plt.title('Promotion Gap Across Career Clusters')
plt.xlabel('Career Cluster')
plt.ylabel('Average Years Since Last Promotion')

plt.show()
```

```python
# STEP 10: KPI Analysis

kpi = df.groupby('Career_Cluster').agg(
    Employee_Count=('Career_Cluster', 'size'),
    Avg_Promotion_Gap=('YearsSinceLastPromotion', 'mean'),
    Avg_Role_Stagnation=('YearsInCurrentRole', 'mean'),
    Avg_Training=('TrainingTimesLastYear', 'mean'),
    Avg_Company_Experience=('YearsAtCompany', 'mean')
).round(2)

# Calculate KPI scores
kpi['Promotion_Gap_Score'] = (
    kpi['Avg_Promotion_Gap'] / df['YearsSinceLastPromotion'].max() * 100
).round(2)

kpi['Role_Stagnation_Index'] = (
    kpi['Avg_Role_Stagnation'] / df['YearsInCurrentRole'].max() * 100
).round(2)

kpi['Training_Intensity_Score'] = (
    kpi['Avg_Training'] / df['TrainingTimesLastYear'].max() * 100
).round(2)

print("CAREER PROGRESSION KPI REPORT")
display(kpi)
```

```python
# STEP 11: KPI Visualization

import matplotlib.pyplot as plt
import seaborn as sns

# 1. Promotion Gap Score
plt.figure(figsize=(8,5))
sns.barplot(
    x=kpi.index,
    y=kpi['Promotion_Gap_Score']
)
plt.title('Promotion Gap Score by Career Cluster')
plt.xlabel('Career Cluster')
plt.ylabel('Promotion Gap Score (%)')
plt.show()

# 2. Role Stagnation Index
plt.figure(figsize=(8,5))
sns.barplot(
    x=kpi.index,
    y=kpi['Role_Stagnation_Index']
)
plt.title('Role Stagnation Index by Career Cluster')
plt.xlabel('Career Cluster')
plt.ylabel('Stagnation Index (%)')
plt.show()

# 3. Training Intensity Score
plt.figure(figsize=(8,5))
sns.barplot(
    x=kpi.index,
    y=kpi['Training_Intensity_Score']
)
plt.title('Training Intensity by Career Cluster')
plt.xlabel('Career Cluster')
plt.ylabel('Training Intensity Score (%)')
plt.show()
```
